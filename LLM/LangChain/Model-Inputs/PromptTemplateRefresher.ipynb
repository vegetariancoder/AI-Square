```python
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(
    model="gpt-4",
    temperature=0,
    streaming=True,
    max_tokens=200,
    seed=365
)
```

```python
template = """
    Given the information of {information}. Tell me about two things:
    1. A short summary of the person.
     2. A short summary of the person's family.
"""
```

```python
prompt_temp = PromptTemplate(template=template,input_variables=["information"])
```

```python
prompt_temp.invoke(input={'information': "Elon Musk"})
```

```python
chain = prompt_temp | llm
```

```python
response = chain.invoke(input={"information": "Elon Musk"})
```

```python
print(response.content)
```
