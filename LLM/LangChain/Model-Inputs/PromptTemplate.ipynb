```python
from langchain.prompts import PromptTemplate
```

```python
TEMPLATE = """
System : {description}


Human : Can you tell me the current prime minister of {country}?
"""
```

```python
prompt_template = PromptTemplate(template=TEMPLATE)
```

```python
prompt_template
```

```python
prompt_value = prompt_template.invoke({
    'description': 'You are a genius who know everything about politics.',
    'country': 'India'
})
```

```python
print(prompt_value.text)
```
