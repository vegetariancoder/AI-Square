### Import Libraries

```python
import openai
import os
from langchain_openai.chat_models import ChatOpenAI
```

### Create a Chat

```python
chat = ChatOpenAI(
    model="gpt-4",
    temperature=0,
    streaming=True,
    max_tokens=200,
    seed=365
)
```

```python
response = chat.invoke(''' I have recently blessed with a baby girl, could you suggest some names ''')
```

```python
print(response.content)
```
