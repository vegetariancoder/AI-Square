### Import Libraries

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.prompts import HumanMessagePromptTemplate, SystemMessagePromptTemplate, ChatPromptTemplate
```

```python
chat = ChatOpenAI(
    model = "gpt-4",
    temperature=0,
    streaming=True,
    seed=365,
    max_tokens=100
)
```

### Define the Template

```python
TEMPLATE_S = """
System : {description}
"""

TEMPLATE_H = """
Human : Can you tell me the current prime minister of {country}?
"""
```

```python
system_p = SystemMessagePromptTemplate.from_template(template=TEMPLATE_S)
human_p = HumanMessagePromptTemplate.from_template(template=TEMPLATE_H)
```

```python
chat_template = ChatPromptTemplate.from_messages([system_p, human_p])
```

```python
print(chat_template)
```

```python
chat_value = chat_template.invoke({
    'description': 'You are a genius who know everything about politics.',
    'country': 'Russia'
})
```

```python
response = chat.invoke(chat_value)
```
