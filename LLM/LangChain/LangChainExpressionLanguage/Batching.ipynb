```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
chat = ChatOpenAI(model_name = 'gpt-4',
                  model_kwargs = {'seed':365},
                  temperature = 0,
                  max_tokens = 500)
```

```python
chat_template = ChatPromptTemplate.from_messages([
    ("human","I have recently bought a {racket}, can you suggest some {nameType} for it")
])
```

```python
chain = chat_template | chat
```

```python
%%time
response = chain.invoke({'racket': 'yonex', 'nameType': 'sarcastic'})
```

```python
print(response.content)
```

```python
chain.batch([{ 'racket': 'yonex', 'nameType': 'sarcastic'},
             {'racket': 'victor', 'nameType': 'funny'}])
```
