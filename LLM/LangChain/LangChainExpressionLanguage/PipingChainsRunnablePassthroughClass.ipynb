```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
chat_template_sport = ChatPromptTemplate.from_template(
    '''
    What are the top 5 skills require to be good at {sports} ?
    Answer by listing in skills
    '''
)

chat_template_strategy = ChatPromptTemplate.from_template(
    '''
    Consider this skills provided, how can master these skills ?
    {skills}
    '''
)
```

```python
chat = ChatOpenAI(model_name = 'gpt-4',
                  model_kwargs = {'seed':365},
                  temperature = 0,
                  max_tokens = 500,
                  streaming=True)
```

```python
str_parser = StrOutputParser()
```

```python
chain_sports = chat_template_sport | chat | str_parser | {'skills':RunnablePassthrough()}
```

```python
chain_strategy = chat_template_strategy | chat | str_parser
```

```python
chain_combined = chain_sports | chain_strategy
```

```python
chain_combined.invoke({'sports': 'badminton'})
```
