### Import Libraries

```python
import openai
import os
```

```python
openai.api_key = os.getenv('OPENAI_API_KEY')
```

### Setting Client

```python
client = openai.OpenAI()
```

### Chat Completion

```python
chat = client.chat.completions.create(
    model="gpt-4",
    messages=[
        {"role": "system",
        "content": "You are a badminton coach who talks badminton only."
         },
        {"role": "user",
         "content": "Who is peter gade ?"}
    ],
    temperature=0,
    max_tokens=50
)
```

```python
chat.choices[0].message.content
```

### Seed, Temperature and MaxTokens

```python
chat = client.chat.completions.create(
    model="gpt-4",
    messages=[
        {"role": "system",
        "content": "You are a badminton coach who talks badminton only."
         },
        {"role": "user",
         "content": "Who is peter gade ?"}
    ],
    temperature=0,
    max_tokens=200,
    seed=356,
    stream=True
)
```

```python
for i in chat:
    print(i.choices[0].delta.content, end="")
```
