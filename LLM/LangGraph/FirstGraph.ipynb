### Import Libraries

```python
from langchain_core.messages import HumanMessage, BaseMessage
from langchain_openai import ChatOpenAI
from collections.abc import Sequence
from langchain_core.runnables import Runnable
from langgraph.graph import START, END , StateGraph
from typing_extensions import TypedDict
from typing import Literal
```

### Define State

```python
class State(TypedDict):
    messages : Sequence[BaseMessage]
```

```python
state = State(messages=[HumanMessage(content="What is the fastest way to learn Badminton?")])
```

```python
state['messages'][0].pretty_print()
```

### Define Node

```python
chat = ChatOpenAI(model_name = 'gpt-4',
                  model_kwargs = {'seed':365},
                  temperature = 0,
                  max_tokens = 500)
```

```python
response = chat.invoke(state['messages'])
```

```python
response.pretty_print()
```

```python
def chatbot(state: State) -> State:

    print(f"\n------> Entering the Chatbot : <-----\n")

    response = chat.invoke(state['messages'])
    response.pretty_print()
    return State(messages=[response])
```

```python
chatbot(state)
```

### Define Graph

```python
graph = StateGraph(State)
```

```python
graph.add_node("chatbot",chatbot)
graph.add_edge(START,"chatbot")
graph.add_edge("chatbot",END)
```

```python
graph_compiled = graph.compile()
```

```python
isinstance(graph_compiled,Runnable)
```

```python
graph_compiled
```

```python
graph_compiled.invoke(state)
```
