### Pre-Built Invoice Processing with Azure Document Intelligence

#### Setting up the Environment

```python
import os
from dotenv import load_dotenv
from azure.ai.formrecognizer import DocumentAnalysisClient
from azure.core.credentials import AzureKeyCredential

load_dotenv()

endpoint = os.getenv("FOUNDRY_PROJECT_ENDPOINT")
key = os.getenv("FOUNDRY_KEY")
```

### Creating the Document Intelligence Client

```python
document_analysis_client=DocumentAnalysisClient(endpoint=endpoint,
                                                credential=AzureKeyCredential(key))
```

### Analyzing an Invoice Document from URL

```python
document_url = "https://github.com/kuljotSB/AI-102-Certification/raw/refs/heads/main/Information-Extraction/Document-Intelligence/Pre-Built-Offerings/Samples/sample-invoice.pdf"
```

```python
poller = document_analysis_client.begin_analyze_document_from_url("prebuilt-invoice", document_url)
classify = document_analysis_client.begin_classify_document_from_url("prebuilt-invoice", document_url)
invoices=poller.result()
invoices_classify=classify.result()
```

```python
for idx,invoice in enumerate(invoices.documents):
    vendor_name=invoice.fields.get("VendorName")
    vendor_name_value=vendor_name.value

    vendor_address = invoice.fields.get("VendorAddress")
    house_number = vendor_address.value.house_number
    street_address = vendor_address.value.street_address
    city = vendor_address.value.city
    state = vendor_address.value.state
    postal_code = vendor_address.value.postal_code

    invoice_date=invoice.fields.get("InvoiceDate")
    invoice_date_value = invoice_date.value

    subtotal=invoice.fields.get("SubTotal")
    subtotal_value=subtotal.value

#printing the extracted fields
print('vendor_name:',vendor_name_value)
print('vendor_address:',house_number + " " + street_address + ", " + city + ", " + state + " " + postal_code)
print('invoice_date:',invoice_date_value)
print('subtotal:',subtotal_value)

```
