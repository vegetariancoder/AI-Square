```python
from langchain_community.document_loaders import Docx2txtLoader
from langchain_text_splitters import MarkdownHeaderTextSplitter
```

```python
doc_reader = Docx2txtLoader("/Users/sahilnagpal/Desktop/AI-Square/RAG/data/Introduction_to_Data_and_Data_Science.docx")
```

```python
doc_data = doc_reader.load()
```

```python
markdown_split = MarkdownHeaderTextSplitter(headers_to_split_on=[
    ('#','Course'),
    ('##','Lecture')
])
```

```python
split_doc_data = markdown_split.split_text(doc_data[0].page_content)
```

```python
for content in split_doc_data:
    content.page_content = content.page_content.replace('\n', ' ')
```
