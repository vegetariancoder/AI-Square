```python
from langchain_community.vectorstores import Chroma
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_core.documents import Document
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-ada-002")
```

```python
vector_store = Chroma(
    persist_directory="/Users/sahilnagpal/Desktop/AI-Square/RAG/intro-data-science-vectorstore",
    embedding_function=embeddings
)
```

```python
retrived_docs = vector_store.similarity_search(query="What software does data scientist use?",
                               k=5)

for content in retrived_docs:
    print(content.page_content)
```
