```python
from langchain_community.document_loaders import Docx2txtLoader
from langchain_text_splitters import MarkdownHeaderTextSplitter
from langchain_text_splitters import CharacterTextSplitter
from langchain.embeddings import OpenAIEmbeddings
```

```python
doc_reader = Docx2txtLoader("/Users/sahilnagpal/Desktop/AI-Square/RAG/data/Introduction_to_Data_and_Data_Science.docx")
```

```python
doc_data = doc_reader.load()
```

```python
# markdown header splitter
md_split = MarkdownHeaderTextSplitter(
    headers_to_split_on=[
        ("#","Course"),
        ("##","Lecture")
    ]
)

md_split_doc_data = md_split.split_text(doc_data[0].page_content)

# remove new line

for content in md_split_doc_data:
    content.page_content = content.page_content.replace("\n"," ")
```

```python
# character splitter
ch_split = CharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=0,
    separator="."
)

ch_split_doc_data = ch_split.split_documents(md_split_doc_data)
```

```python
embedding = OpenAIEmbeddings(model="text-embedding-ada-002")
```

```python
vector1 = embedding.embed_query(ch_split_doc_data[0].page_content)
vector1
```
