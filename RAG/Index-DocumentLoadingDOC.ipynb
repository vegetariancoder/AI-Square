```python
from langchain_community.document_loaders import Docx2txtLoader
from langchain_text_splitters import CharacterTextSplitter
```

```python
loader_doc = Docx2txtLoader("/RAG/data/Introduction_to_Data_and_Data_Science.docx")
```

```python
pages_doc = loader_doc.load()
```

#### Before remoing characters

```python
len(pages_doc[0].page_content)
```

#### Remove New Line Character

```python
for content in range(len(pages_doc)):
    pages_doc[content].page_content = " ".join(pages_doc[content].page_content.split())
```

#### After removing characters

```python
len(pages_doc[0].page_content)
```

```python
text_split = CharacterTextSplitter(separator=".",
                                   chunk_size=500,
                                   chunk_overlap=50)
```

```python
pages_char_split = text_split.split_documents(pages_doc)
```
