```python
from langchain_core.documents import Document
from langchain_community.embeddings import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
```

```python
embedding = OpenAIEmbeddings(model="text-embedding-ada-002")
```

```python
vector_store = Chroma(persist_directory="/Users/sahilnagpal/Desktop/AI-Square/RAG/intro-data-science-vectorstore/", embedding_function=embedding)
```

```python
len(vector_store.get()['documents'])
```

```python
retriever = vector_store.as_retriever(
    search_type="mmr",
    search_kwargs={"k": 3,
                   'lambda_mult':0.7
                   }
)
```

```python
retrieved_docs = retriever.invoke("What software does data scientist used ?")
```

```python
for data in retrieved_docs:
    print(data.page_content)
```
