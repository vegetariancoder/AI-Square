#### Import Library

```python
from langchain_core.prompts import ChatPromptTemplate
from Project.Customer_Support_Ticket_Analyser.config import get_chat_model
```

#### Create REPLY PROMPT from ChatPromptTemplate

```python
reply_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            """
You are a helpful customer support agent.

Write polite and professional responses to customer tickets.
"""
        ),

        # Few-shot example
        (
            "human",
            "Ticket: I was charged twice for my subscription."
        ),
        (
            "ai",
            "We apologize for the duplicate charge. Our billing team will review your account and issue a refund within 24 hours."
        ),

        (
            "human",
            "Ticket: I cannot login to my account."
        ),
        (
            "ai",
            "We are sorry you are experiencing login issues. Please try resetting your password using the password recovery option."
        ),

        ("human", "Ticket: {ticket}")
    ]
)
```

```python
print(reply_prompt.messages[0].prompt.template)
```

#### Invoke Chat Prompt Template

```python
ticket_dict = {
    'ticket': 'My credit card was charged twice'
}
```

```python
chat_template = reply_prompt.invoke(ticket_dict)
```

#### Invoke Chat to Classify System Message

```python
chat = get_chat_model()
```

```python
response = chat.invoke(chat_template)
```

### Create the Chain of Prompts

```python
chain = reply_prompt | chat
result = chain.invoke(ticket_dict)
```
